```python
from sklearn.model_selection import GroupShuffleSplit

# -------------------------------
# GroupShuffleSplit (80:20)
# -------------------------------

gss = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, test_idx = next(
    gss.split(
        df58,
        groups=df58["Participant_ID"]
    )
)

train_df = df58.iloc[train_idx].reset_index(drop=True)
test_df  = df58.iloc[test_idx].reset_index(drop=True)

# -------------------------------
# Shapes
# -------------------------------

print("Original Dataset :", df58.shape)
print("Training Dataset :", train_df.shape)
print("Testing Dataset  :", test_df.shape)

# -------------------------------
# Participant counts
# -------------------------------

train_participants = set(train_df["Participant_ID"])
test_participants  = set(test_df["Participant_ID"])

print("\nUnique Participants in Train :", len(train_participants))
print("Unique Participants in Test  :", len(test_participants))

# -------------------------------
# Leakage Check
# -------------------------------

overlap = train_participants.intersection(test_participants)

print("\nNumber of overlapping participants :", len(overlap))

if len(overlap) == 0:
    print("No participant leakage detected.")
else:
    print("Leakage detected!")
    print(overlap)
```

```python
train_df["Status"] = df58.iloc[train_idx]["Status"].values
test_df["Status"]  = df58.iloc[test_idx]["Status"].values
```

```python
print(train_df.columns)
print(test_df.columns)
```

```python
import pandas as pd
import numpy as np
import shap
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
# ============================================================
# Separate Features and Labels
# ============================================================

X_train = train_df.drop(columns=["Participant_ID", "Status"])
y_train = train_df["Status"]

X_test = test_df.drop(columns=["Participant_ID", "Status"])
y_test = test_df["Status"]

# ============================================================
# Standardization
# (Fit ONLY on training data to avoid leakage)
# ============================================================

scaler = StandardScaler()

X_train_scaled = pd.DataFrame(
    scaler.fit_transform(X_train),
    columns=X_train.columns,
    index=X_train.index
)

X_test_scaled = pd.DataFrame(
    scaler.transform(X_test),
    columns=X_test.columns,
    index=X_test.index
)

# ============================================================
# Random Forest (TRAIN ONLY)
# ============================================================

rf = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

rf.fit(X_train_scaled, y_train)

# ============================================================
# SHAP Feature Importance (TRAIN ONLY)
# ============================================================

# Use SHAP values for the PD class (class index = 1)
explainer = shap.TreeExplainer(rf)
shap_values = explainer.shap_values(X_train_scaled)

# Mean absolute SHAP importance for PD class
shap_importance = np.abs(shap_values[:, :, 1]).mean(axis=0)
# ============================================================
# Top 7 Features
# ============================================================

feature_importance = pd.DataFrame({
    "Feature": X_train_scaled.columns,
    "Importance": shap_importance
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

top7_features = feature_importance.head(7)["Feature"].tolist()

```

```python
# ============================================================
# Keep ONLY Top 7 Features
# ============================================================

train_top7 = train_df[["Participant_ID", "Status"] + top7_features].copy()
test_top7  = test_df[["Participant_ID", "Status"] + top7_features].copy()

print("\nTraining Shape after RF-SHAP :", train_top7.shape)
print("Testing Shape after RF-SHAP  :", test_top7.shape)

# ============================================================
# Combine Back
# ============================================================

df58_rfshap = pd.concat(
    [train_top7, test_top7],
    ignore_index=True
)

print("\nFinal Dataset Shape :", df58_rfshap.shape)
```
